1. Swap an Array

```python


arr=list(map(int,input("Enter the elements separated by space: ").split()))

l=0
r=len(arr)-1

while l<r:
    arr[l],arr[r]=arr[r],arr[l]

    l+=1
    r-=1

print(f"{arr}")
```

2. Find Max and Min

```python


arr=list(map(int,input("Enter the no. u want to add to the list: ").split()))

min=arr[0]
max=arr[0]

for i in arr:
    if i<min:
        min=i
    if i>max:
        max=i
print(f"Maximum no. in list is: {max}\nMinimum no. in list is: {min}")
```

3. Find the 3rd Largest Number

```python


arr=list(map(int,input("Enter the elements u want to add: ").split()))

x=(float('-inf'))
y=(float('-inf'))
z=(float('-inf'))

for i in arr:
    if i>x:
        z=y
        y=x
        x=i
    elif i>y and i!=x:
        z=y
        y=i
    elif i>z and i!=y and i!=x:
        z=i
if z==(float('-inf'))
    print("Invalid")
else:
    print(f"Third largeat no. is: {z}")
```

4. Move all zeroes to end

6. Check sorted or not

```python


def sort(ar):
    for i in range(len(ar)-1):
        if ar[i]>ar[i+1]:
            return False
    return True

arr=list(map(int,input("Enter your elements: ").split()))
if sort(arr)==True:
    print("Entered array is 'sorted'")
else:
    print("Entered array is 'not sorted'")
```

7. First and Last index of a number

```python


arr=list(map(int,input("Enter your elements: ").split()))
x=int(input("Enter the targeted no.: "))
first=-1
last=-1

for i in range (len(arr)):
    if arr[i]==x:
        if first == -1:
            first=i
        last=i

if first!=-1:
    print(f"First index:{first}\nLast index:{last}")
else:
    print("Not found")
```

8. Remove duplicates

```python


arr=list(map(int,input("Enter your elements: ").split()))
if len(arr)>0:
    j=0
    for i in range (1,len(arr)):
        if arr[i]!=arr[j]:
            j+=1
            arr[j]=arr[i]

    arr=arr[:j+1]
print(f"{arr}")
```

10. Unique Element in an Array

```python


arr=list(map(int,input("Enter your elements: ").split()))
count={}
for i in arr:
    if i in count:
        count[i]+=1
    else:
        count[i]=1
result=[]
for i in count:
    if count [i]==1:
        result.append(i)
print(f"Unique element: {result}")
```

11. Find 2 numbers from whose sum is the targeted input number

```python


arr=list(map(int,input("Enter your elements: ").split()))
trgt=int(input("Enter the targeted no.: "))

i=0
j=len(arr)-1
x=[]
while i<j:
    sum=arr[i]+arr[j]
    if sum == trgt:
        x.append((i,j))
        i+=1
        j-=1
    elif sum<trgt:
        i+=1
    else:
        j-=1

print(f"INDEX{x}")
```

12. Find the highest possible area of the rectangle using 2 numbers from the array

```python


arr=list(map(int,input("Enter the elements u want to add: ").split()))

j=len(arr)-1
i=0
x=0

while i<j:

    if arr[i]<=arr[j]:
        if (j-i)*arr[i]>x:
            x=(j-i)*arr[i]
        i+=1
    else:
        if(j-i)*arr[j]>x:
            x=(j-i)*arr[j]
        j-=1

print(f"{x}")
```

13. Find the Trapped Water

```python


arr=list(map(int,input("Enter the elements u want to add: ").split()))
i=0
j=(len(arr)-1)
k=0
l=0
x=0
while i<j:

    if arr[i]>k:
        k=arr[i]
    if arr[j]>l:
        l=arr[j]
    if k>l:
        x+=l-arr[j]
        j-=1

    else:
        x+=k-arr[i]
        i+=1


print(f"{x}")
```
